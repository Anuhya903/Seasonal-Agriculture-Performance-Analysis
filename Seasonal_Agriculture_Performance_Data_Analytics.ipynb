## Outlier Analysis

Outlier analysis is performed to identify unusually high or low observations
in the agricultural data. These observations are investigated because they
may represent genuine differences between farms, crops, seasons or farming
practices, or they may indicate possible data quality issues.

```python
# Descriptive statistics for Yield

print("Mean Yield:", df['Yield_Tonnes_Ha'].mean())
print("Median Yield:", df['Yield_Tonnes_Ha'].median())
print("Standard Deviation:", df['Yield_Tonnes_Ha'].std())
print("Minimum Yield:", df['Yield_Tonnes_Ha'].min())
print("Maximum Yield:", df['Yield_Tonnes_Ha'].max())
```

### IQR Method for Detecting Yield Outliers

The Interquartile Range (IQR) method is used to identify potential
outliers in Yield.

Q1 represents the 25th percentile and Q3 represents the 75th percentile.
The IQR is the difference between Q3 and Q1.

Observations below Q1 - 1.5 × IQR or above Q3 + 1.5 × IQR are considered
potential outliers.

```python
# Calculate IQR for Yield

Q1 = df['Yield_Tonnes_Ha'].quantile(0.25)
Q3 = df['Yield_Tonnes_Ha'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - (1.5 * IQR)
upper_bound = Q3 + (1.5 * IQR)

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
yield_outliers = df[
    (df['Yield_Tonnes_Ha'] < lower_bound) |
    (df['Yield_Tonnes_Ha'] > upper_bound)
]

print("Number of yield outliers:", len(yield_outliers))
```

variables. For example, water efficiency is calculated using yield and

# Recommendations

The following recommendations are based on the patterns and relationships
observed during the analysis. They are intended to support better seasonal
agricultural planning and resource management.

### Recommendation 1 — Focus on profitable seasonal opportunities

Agricultural planning could give greater attention to profitable
crop-season combinations observed during the Kharif season, while
considering associated risks and regional suitability. The analysis
shows that Kharif has substantially higher average profitability than
Rabi and Zaid.

### Recommendation 2 — Investigate the causes of low Zaid profitability

The negative average profit observed during the Zaid season suggests
that its cost structure, crop selection, revenue generation and
resource requirements should be investigated. Identifying the factors
responsible for the lower profitability could help improve seasonal
planning and decision-making.

### Recommendation 3 — Evaluate water usage against agricultural output

Water management should be evaluated by considering both water
consumption and the agricultural output obtained from it. Higher water
usage does not necessarily indicate higher yield, so planning should
focus on how efficiently available water is converted into agricultural
output.

### Recommendation 4 — Evaluate irrigation methods using multiple measures

Irrigation methods should be evaluated using yield, profitability,
water usage and water efficiency together rather than relying on a
single performance measure. The analysis shows that Drip irrigation
has stronger average yield and profitability, while Rainfed agriculture
has higher water efficiency.

### Recommendation 5 — Strengthen pest and disease monitoring

Kharif records the highest observed average disease and pest risk at
54.47%. Therefore, greater attention could be given to disease and
pest monitoring during Kharif to help manage the higher observed risk.

### Recommendation 6 — Consider crop-season combinations in planning

Crop selection should be considered together with season rather than
evaluating crops independently. The analysis identified Sugarcane
during Kharif as a particularly strong crop-season combination in
terms of observed average yield and profit. Such combinations can be
considered for further evaluation while accounting for water needs,
costs, market conditions and regional suitability.

### Recommendation 7 — Use region-specific seasonal planning

Seasonal agricultural strategies should consider regional differences.
The State-Season analysis indicates that agricultural performance can
vary across both location and season. Therefore, a single seasonal
strategy may not be equally effective across all regions.

### Recommendation 8 — Consider multiple factors when planning yield

The correlation analysis shows very weak linear relationships between
rainfall, average temperature, soil moisture and yield in the available
data. Therefore, agricultural planning should not rely on these
environmental variables alone. Crop type, irrigation, resource usage,
production practices and economic factors should also be considered.

## Limitations

1. Missing values were present in some variables and were handled using
   season-wise median imputation.

2. The number of observations differs across seasons, which may affect
   the comparability of seasonal results.

3. Extreme observations may influence averages and correlations.

4. Correlation shows association but does not prove causation.

5. Seasonal differences may be influenced by several factors at the same
   time, including crop selection, environmental conditions, irrigation,
   costs and market conditions.

6. The dataset alone cannot establish that a particular agricultural
   practice directly causes higher yield or profit.

7. Additional information such as detailed weather history, market price
   fluctuations and farm-management practices could improve future
   analysis.

8. The dataset contains different crops with substantially different
   yield ranges, so overall comparisons and outlier detection may be
   influenced by crop composition.

## Conclusion

This project analyzed agricultural performance across Kharif, Rabi and
Zaid seasons using environmental, resource, production and economic
variables.

The analysis found that Kharif recorded the highest average yield of
5.63 tonnes/ha and the highest average profit of ₹178,914.65, while
Zaid recorded the lowest average yield of 4.64 tonnes/ha and a negative
average profit of ₹24,804.82. Zaid also had the highest average water
usage, whereas Kharif showed the highest average water efficiency.

The crop-season analysis identified Sugarcane-Kharif as the strongest
observed crop-season combination in terms of average yield and profit.
The irrigation analysis showed that Drip irrigation had the highest
average yield and profit, while Rainfed agriculture had the highest
water efficiency.

Environmental variables such as rainfall, temperature and soil
moisture showed very weak linear relationships with yield in the
available data. This suggests that yield differences cannot be
explained by these environmental variables alone and may involve
multiple agricultural and management factors.

Overall, the findings suggest that seasonal agricultural planning should
consider profitability, productivity, water efficiency, irrigation
method, crop-season combinations, regional differences and seasonal
risk together rather than relying on a single performance measure.

However, these findings represent observed patterns and associations
within the available dataset and should not be interpreted as proof of
causal relationships.
